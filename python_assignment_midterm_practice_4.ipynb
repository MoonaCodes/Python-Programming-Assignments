```python
print("=" * 64)
print("VARIABLES & DATA TYPES")
print("=" * 64)

print("\nQuestion:1")
print("Write a Python program that stores a student's name, age, marks, and pass status in four variables.\n"
"Print all values and use type() to display the data type of each variable.")

print("\n")
print("Students name,age,marks and pass status:")
name = "Moiz"
age = 26
marks = 78.5
pass_status = False


print("Name",name)
print("age",age)
print("Marks",marks)
print("Pass status",pass_status)

print("\n")
print("All Values")
print(name)
print(age)
print(marks)
print(pass_status)

print("\n")
print("Data Types:")
print("Name",type(name))
print("Age",type(age))
print("Marks",type(marks))
print("Pass Status",type(pass_status))
```
