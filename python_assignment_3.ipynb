```python
print("=" * 64)
print("BONUS PRACTICE")
print("=" * 64)

print("\nBonus:")
print("Create a list [1,2,3,2,1]. Make a copy of the list, reverse the copy,\nand compare it with the original list to check whether it is a palindrome.")

# Create a list

stats = [1,2,3,2,1]
print("\nOriginal copy: ",stats)

# Copy of the list
print("\nCopied list")
gain = stats.copy()
print(gain)

# Reverse of the list
print("\nReversed list: ")
gain.reverse()
print(gain)

# Comparison of list
print("\nComparison: ")
if stats == gain:
  print("It is a Palindrome")
else:
  print("It is not a Palindrome")

print("\n" + "-" * 64)
print("Program executed successfully")
print("-" * 64)
```
